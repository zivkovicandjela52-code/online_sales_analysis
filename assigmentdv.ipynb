## Korak 1 : import podataka

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv(
    "C:/Users/PC/Desktop/data analysis/netflix_dataset.csv",
    encoding="latin1",
    engine="python",
    on_bad_lines="skip"
)

column_names = df.columns.to_list()

print("Nazivi kolona su:", column_names)
print(df.dtypes)
```

## KORAK 2
SREDI PODATKE DA MOGU DA UDJU U ANALIZU

```python
df["release_year"] = pd.to_numeric(df["release_year"], errors="coerce")

df["genres"] = df["genres"].str.replace("[", "", regex=False)
df["genres"] = df["genres"].str.replace("]", "", regex=False)
df["genres"] = df["genres"].str.replace("'", "", regex=False)

# razdvoji žanrove po zarezu
df["genres"] = df["genres"].str.split(",")

# napravi novu tabelu gde svaki žanr ima svoj red
df_genres = df.explode("genres")
```

## trend 1
ispitujemo trend imdb-a score-a u odnosu na godinu.

```python
rating_trend = df.groupby("release_year")["imdb_score"].mean().reset_index()

plt.figure()
sns.lineplot(data=rating_trend, x="release_year", y="imdb_score")

plt.title("Prosečna IMDb ocena kroz vreme")
plt.xlabel("Godina")
plt.ylabel("IMDb score")
plt.xlim(1945, 2022)
plt.xticks(range(1945, 2023, 10))
plt.show()
```

## analiza
Na grafiku vidimo da se ocene krecu od 4.5 do 8.5 . a na x osi vidimo vremensku raspodelu od 1945 do 2015+.  Vidimo da je trend imdb score-a najvisi sa starim klasicnim filmovima od 40 tih do 60tih. Prosek je uglavnom izmedju 7 i 8.5 sa jednim izrazajnim pikom. Zatim imamo trend pada na 5.5, oko 70ih i 80ih  i kasnije krece da raste ali ne znacajno i filmovi nikada ne dostizu kvalitet kao u ranijim godinama sa pocetkom filmske industrije, ali imaju odredjenu stabilnost. Sa blagim padom od 2010 .

## trend 2
Ispitujemo koja varijable imaju najveci uticaj jedna na drugu

```python

corr = df[["runtime","imdb_score","imdb_votes","tmdb_popularity","tmdb_score"]].corr()

plt.figure()
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Korelacija između ključnih metrika")
plt.show()
```

## analiza
iz matrice korelacije mozemo da vidimo kako promenljive medjusobno uticu jedna na drugu. To je vrlo znacajna polazna tacka da znamo sta bismo mogli da analiziramo. Pokazuje kolko su u koleraciji brojcano i bojom . Imdb score ima jaku pozitivnu korelaciju sa tmbd scorom i pokazuje negativnu korelaciju sa duzinom trajanja -0.21. Popularnost pokazuje pozitivnu korelaciju sa brojem glasova, sto je i logicno. Imdb score i broj glasova pokazuju pozitivnu korelaciju 0.19.
Postoji pozitivna veza izmedju IMDB I TMDB ocena, sto znaci da postoji stabilnost u percepciji ocene kvaliteta na obe platforme. Duzina trajanja iako negativo korelise, ne predstavlja znacajnu korelaciju. Popularnost i ocena nisu znacajno povezani.
